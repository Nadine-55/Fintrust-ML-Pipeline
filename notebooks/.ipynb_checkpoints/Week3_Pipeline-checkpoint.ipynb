```python
#Feature building
from src.preprocessing import build_features

train_df = build_features(transactions, customers, mode='train')
print(train_df.shape)

new_txn = transactions.drop(columns=['Risk_Review_Flag']).head(1)
live_df = build_features(new_txn, customers, mode='inference')
print(live_df.shape)
```

```python
#Model Building

from sklearn.metrics import classification_report
from src.model import time_based_split, build_pipeline, get_target, FEATURE_COLUMNS

features_df = build_features(transactions, customers, mode='train')
train_part, test_part = time_based_split(features_df)

X_train, y_train = train_part[FEATURE_COLUMNS], get_target(train_part)
X_test, y_test = test_part[FEATURE_COLUMNS], get_target(test_part)

pipe = build_pipeline()
pipe.fit(X_train, y_train)
print(classification_report(y_test, pipe.predict(X_test)))

# Demo case: one live transaction
print(pipe.predict(live_df[FEATURE_COLUMNS]))
```

```python
#Save and Load Model

import numpy as np
from src.model import save_model, load_model

saved_path = save_model(pipe)
print(saved_path)

loaded = load_model()
print(np.array_equal(pipe.predict(X_test), loaded.predict(X_test)))
```

```python
from sklearn.ensemble import RandomForestClassifier
from src.preprocessing import build_features
from src.model import time_based_split, build_pipeline, get_target, FEATURE_COLUMNS
from src.pipeline import run_prediction

features_df = build_features(transactions, customers, mode='train')
train_part, test_part = time_based_split(features_df)
X_train, y_train = train_part[FEATURE_COLUMNS], get_target(train_part)

alt = build_pipeline()
alt.set_params(classifier=RandomForestClassifier(n_estimators=100, random_state=42,
                                                 class_weight='balanced'))
alt.fit(X_train, y_train)

alt_scores, _ = run_prediction(transactions.head(5), customers, model=alt)
print(alt_scores)
```
